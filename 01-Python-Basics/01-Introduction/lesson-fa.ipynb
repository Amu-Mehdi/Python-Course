## ۱) برنامه‌نویسی چیه؟

فرض کن توی یه رستوران نشستی. تو سفارش می‌دی، گارسون سفارش رو به آشپز می‌رسونه و آشپز غذا رو آماده می‌کنه. اگه سفارشت رو درست و واضح بگی، همون غذایی که می‌خوای می‌یاد.

برنامه‌نویسی (programming) دقیقاً همینه: **ما دستور می‌نویسیم و کامپیوتر اون دستور رو انجام می‌ده.**

**چرا لازمه؟** چون کامپیوتر خودش هیچ کاری رو نمی‌دونه؛ فقط دستور اجرا می‌کنه. هر چیزی که روی گوشی و کامپیوتر می‌بینی (بازی، پیام‌رسان، ماشین‌حساب) یه مجموعه دستور بوده که یه برنامه‌نویس نوشته.

**دستورهایی که می‌نویسیم اسمش «کد» (code) و چیزی که کامپیوتر نشون می‌ده اسمش «خروجی» (output) هست:**

```text
برنامه‌نویس (تو)
      │  دستور (کد) می‌نویسه
      ▼
   پایتون
      │  دستور رو می‌فهمه و اجرا می‌کنه
      ▼
  کامپیوتر
      │  کار رو انجام می‌ده
      ▼
  خروجی (نتیجه روی صفحه)
```

## ۲) پایتون چیه؟

پایتون (Python) یه «زبان برنامه‌نویسی» هست؛ یعنی یه زبون مشترک بین تو و کامپیوتر. اون رو یه آدم هلندی به اسم گیدو فان روسوم ساخت.

**کجا استفاده می‌شه؟**

```text
پایتون
├── ساخت وب‌سایت و برنامه‌های اینترنتی
├── تحلیل داده و آمار
├── هوش مصنوعی و یادگیری ماشین
├── خودکارسازی کارهای تکراری
├── شبکه و امنیت
└── ساخت بازی و ابزارهای کوچیک
```

**چرا این‌قدر محبوبه؟**
- نوشتنش ساده و شبیه جمله‌های انگلیسیه، برای همین یادگیریش راحته.
- رایگانه و جامعه جهانی بزرگی داره.
- کتابخونه‌های آماده زیادی داره که کار رو سریع می‌کنن.
- هم برای شروع خوبه، هم برای کارهای حرفه‌ای.

## ۳) کجا و چطوری پایتون رو اجرا کنیم؟

برای نوشتن کد به یه ویرایشگر (editor) نیاز داریم. معروف‌ترینش ویژوال استودیو کد (Visual Studio Code) هست که بهش می‌گیم «وی‌اس‌کد». این‌جا ساختار کلی پنجره‌اش رو ببین:

```text
┌─────────────────────────────────────────────────────────────┐
│ File  Edit  Selection  View  Go  Run  Terminal  Help   (1)  │
├──────────┬──────────────────────────────────────────────────┤
│          │ hello.py                                  (2)    │
│ EXPLORER │                                                  │
│ (3)      │ 1  print("Hello")                       (4)      │
│          │ 2                                                │
│          ├──────────────────────────────────────────────────┤
│          │ TERMINAL                                  (5)    │
│          │ Hello                                            │
└──────────┴──────────────────────────────────────────────────┘
```

- (1) **نوار منو:** منوهای بالای برنامه.
- (2) **تب فایل:** فایلی که الان بازه (hello.py).
- (3) **پنل فایل‌ها:** لیست فایل‌های پروژه.
- (4) **محل نوشتن کد:** این‌جا کد می‌نویسی.
- (5) **ترمینال (Terminal):** این‌جا خروجی برنامه رو می‌بینی.

**دو روش اجرا:**

**روش اول: فایل نوت‌بوک (Notebook).** همین فایلی که الان داری می‌خونی یه نوت‌بوکه. هر تیکه کد توی یه «سلول» (cell) نوشته می‌شه و با زدن دکمه اجرا یا کلید زیر اجرا می‌شه:

```text
Shift + Enter
```

شکل کلی یه سلول کد:

```text
┌─────────────────────────────────────────────────┐
│ [ ]:  print("Hello")                    ▶ Run   │  ← (1) سلول کد
├─────────────────────────────────────────────────┤
│ Hello                                           │  ← (2) خروجی
└─────────────────────────────────────────────────┘
```

اگه برنامه اولین بار ازت پرسید «کدوم هسته (kernel)؟»، گزینه پایتون ۳ (Python 3) رو انتخاب کن.

**روش دوم: فایل معمولی پایتون.** یه فایل با پسوند py. بساز و کدت رو توش بنویس، بعد توی ترمینال این دستور رو بزن:

```text
python hello.py
```
اگه جواب نداد، این رو امتحان کن:

```text
py hello.py
```
برای این‌که ببینی پایتون نصب هست یا نه، این دستور رو توی ترمینال بزن:

```text
python --version
```

## ۴) اولین برنامه

بالاخره وقتش رسید! اولین برنامه‌ات رو بنویس. اول خودت کد رو اجرا کن، بعد توضیحش رو بخون.

````python
print('Python is fun')
````

**کامپیوتر حرف من رو چطور می‌فهمه؟**

گیومه تکی و دوتایی هر دو کار می‌کنن.

```text

```

## ۶) کامنت

کامنت (comment) یه توضیح برای آدم‌هاست، نه برای کامپیوتر. هر چیزی که بعد از علامت `#` بنویسی، پایتون نادیده می‌گیره.

**چرا لازمه؟** چون وقتی کدت بزرگ می‌شه یا چند وقت بعد برمی‌گردی بهش، با کامنت زود می‌فهمی هر قسمت چه کاری می‌کنه.

**کاربرد دومش:** می‌تونی یه خط کد رو موقتاً غیرفعال کنی، بدون این‌که پاکش کنی.

### کامنت بالای خط

**من به کامپیوتر چی می‌گم؟**

«پایتون، خط اول رو کاری بهش نداشته باش؛ فقط خط دوم رو اجرا کن.»

````python
# This line prints a greeting
print("Hi!")
````

**کامپیوتر حرف من رو چطور می‌فهمه؟**

پایتون وقتی # می‌بینه، بقیه اون خط رو نمی‌خونه. پس فقط پرینت اجرا می‌شه.

**در نهایت چه اتفاقی می‌افته؟**

```text
Hi!
```

### کامنت آخر خط

**من به کامپیوتر چی می‌گم؟**

«پایتون، متن Hello رو چاپ کن؛ توضیحی که بعد از # هست مال منه.»

````python
print("Hello")  # say hello
````

**کامپیوتر حرف من رو چطور می‌فهمه؟**

کد قبل از # اجرا می‌شه و بعدش نادیده گرفته می‌شه.

**در نهایت چه اتفاقی می‌افته؟**

```text
Hello
```

### غیرفعال کردن یه خط

**من به کامپیوتر چی می‌گم؟**

«پایتون، فقط خط اول رو اجرا کن؛ خط دوم رو فعلاً کنار گذاشتم.»

````python
print("Visible")
# print("Hidden")
````

**کامپیوتر حرف من رو چطور می‌فهمه؟**

خط دوم با # شروع شده؛ پس پایتون اون رو اصلاً اجرا نمی‌کنه، انگار نیست.

**در نهایت چه اتفاقی می‌افته؟**

```text
Visible
```

برای این‌که یه خط رو سریع کامنت یا از کامنت دربیاری، توی وی‌اس‌کد این میانبر رو بزن:

```text
Ctrl + /
```

## ۷) خطاهای رایج

خطا (error) شکست نیست؛ پایتون داره بهت می‌گه کجا رو اشتباه فهمیده. خطاهای این درس سه تان:

**الف) گیومه بسته نشده**

```python
print("Hello)
```
**من به کامپیوتر چی گفتم؟** «پایتون، متنی رو چاپ کن که تهش رو نگفتم کجا تموم می‌شه!»

**پایتون چی فهمید؟** «شروع متن رو دیدم ولی پایانش رو پیدا نکردم.»

**نتیجه:**

```text
SyntaxError: unterminated string literal (detected at line 1)
```

**ب) پرانتز بسته نشده**

```python
print("Hello"
```
**من به کامپیوتر چی گفتم؟** «دستور رو شروع کردم ولی نگفتم کجا تموم می‌شه.»

**پایتون چی فهمید؟** «یه پرانتز بازه و بسته‌اش رو ندیدم.»

**نتیجه:**

```text
SyntaxError: '(' was never closed
```

**پ) حروف بزرگ و کوچک**

```python
Print("Hello")
```
**من به کامپیوتر چی گفتم؟** «دستوری به اسم Print با P بزرگ.»

**پایتون چی فهمید؟** «من چنین اسمی رو نمی‌شناسم؛ دستورم print با p کوچیکه.»

**نتیجه:**

```text
NameError: name 'Print' is not defined
```
متن دقیق خطا ممکنه بسته به نسخه پایتون کمی فرق داشته باشه، ولی ایده همونه. پیشنهاد می‌کنم خودت هم این کدها رو یه بار امتحان کنی تا خطا رو از نزدیک ببینی.

### تمرین 6 — حل مسئله واقعی: لیست خرید
